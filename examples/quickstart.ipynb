# scAlps · your first landscape

Explore seeded synthetic tissue, then replace `sca.demo()` with your AnnData. All example biology is simulated.

```python
import pyvista as pv

import scalps as sca

pv.set_jupyter_backend("static")  # change to "trame" for interactive exploration
adata = sca.demo()
adata
```

## A mountain in one line

```python
sca.plot(adata, "MKI67", preset="ember")
```

## Cell density and immune neighborhoods

```python
sca.plot(adata, groupby="cell_type", groups="T cell", preset="glacier")
```

## Signed scores form peaks and valleys

```python
sca.plot(adata, "immune_balance", height=0.18)
```

## Build once, export anywhere

Colors represent raw aggregated values. Height is exaggerated and clipped at the absolute 99th percentile by default.

```python
mountain = sca.terrain(adata, "MKI67", resolution=250, smooth=2.5)
mountain.save("outputs/mki67.png", preset="ember")
mountain.save("outputs/mki67.html", preset="ember")
mountain.metadata
```
